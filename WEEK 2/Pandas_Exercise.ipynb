```python
import pandas as pd
import numpy as np

print("--- Intermediate Pandas Exercise ---")
print("Complete each task by writing the requested Pandas code.")
print("------------------------------------")

```

```python
print("\n--- Task 1: Basic Pandas Functions and Converting Arrays to DataFrames ---")

# 1.1 Create a Pandas Series
s1_1 = pd.Series([10, 20, 30, 40, 50])

# 1.2 Create a 3x4 NumPy array with random integers between 1 and 100
np_array1_2 = np.random.randint(1, 101, size=(3, 4))

# 1.3 Convert NumPy array into DataFrame
df1_3 = pd.DataFrame(
    np_array1_2,
    columns=['A', 'B', 'C', 'D']
)

print("\nSeries:")
print(s1_1)

print("\nNumPy Array:")
print(np_array1_2)

print("\nDataFrame:")
print(df1_3)
```

```python
print("\n--- Task 2: Synthetic Data Generation for Practice ---")

# Create DataFrame with 100 rows
df_sales = pd.DataFrame({
    'Date': pd.date_range(start='2023-01-01', end='2023-04-10', periods=100),
    'Region': np.random.choice(
        ['East', 'West', 'North', 'South'],
        size=100
    ),
    'Product': np.random.choice(
        ['Laptop', 'Mouse', 'Keyboard', 'Monitor'],
        size=100
    ),
    'Sales': np.random.randint(100, 1001, size=100),
    'Quantity': np.random.randint(1, 11, size=100)
})

# Introduce NaN values
sales_nan_indices = np.random.choice(df_sales.index, size=5, replace=False)
quantity_nan_indices = np.random.choice(df_sales.index, size=3, replace=False)

df_sales.loc[sales_nan_indices, 'Sales'] = np.nan
df_sales.loc[quantity_nan_indices, 'Quantity'] = np.nan

print("\nSales DataFrame:")
print(df_sales)
```

```python
print("\n--- Task 3: Indexing and Slicing in DataFrames ---")

# 3.1 Select Region and Sales columns
df3_1 = df_sales[['Region', 'Sales']]

# 3.2 Select rows where Product is Laptop
df3_2 = df_sales[df_sales['Product'] == 'Laptop']

# 3.3 Region is East AND Sales > 500
df3_3 = df_sales[
    (df_sales['Region'] == 'East') &
    (df_sales['Sales'] > 500)
]

# 3.4 Select Sales value for first entry where Product is Monitor
monitor_rows = df_sales[df_sales['Product'] == 'Monitor']

if not monitor_rows.empty:
    val3_4 = df_sales.loc[monitor_rows.index[0], 'Sales']
    val3_4 = monitor_rows.iloc[0]['Sales']
else:
    val3_4 = np.nan

print("\nTask 3.1:")
print(df3_1)

print("\nTask 3.2:")
print(df3_2)

print("\nTask 3.3:")
print(df3_3)

print("\nTask 3.4:")
print(val3_4)
```

```python
print("\n--- Task 4: Data Cleaning (Handling Missing Values) ---")

# 4.1 Check missing values in each column
missing_counts4_1 = df_sales.isna().sum()

# 4.2 Fill missing Sales values with mean
df4_2_filled = df_sales.copy()
df4_2_filled['Sales'] = df4_2_filled['Sales'].fillna(
    df4_2_filled['Sales'].mean()
)

# 4.3 Drop rows having any missing values
df4_3_dropped = df_sales.dropna()

print("\nMissing values:")
print(missing_counts4_1)

print("\nDataFrame after filling Sales:")
print(df4_2_filled)

print("\nDataFrame after dropping rows with NaN:")
print(df4_3_dropped)
```

```python
print("\n--- Task 5: Data Manipulation (Columns, Sorting, Grouping) ---")

# 5.1 Create Total_Revenue column
df_sales['Total_Revenue'] = df_sales['Sales'] * df_sales['Quantity']

# 5.2 Sort by Date ascending and Sales descending
df5_2_sorted = df_sales.sort_values(
    by=['Date', 'Sales'],
    ascending=[True, False]
)

# 5.3 Group by Region and calculate sum of Sales and Quantity
df5_3_grouped_region = df_sales.groupby('Region')[['Sales', 'Quantity']].sum()

# 5.4 Group by Product and calculate average Sales
df5_4_grouped_product = df_sales.groupby('Product')['Sales'].mean()

print("\nTask 5.1:")
print(df_sales)

print("\nTask 5.2:")
print(df5_2_sorted)

print("\nTask 5.3:")
print(df5_3_grouped_region)

print("\nTask 5.4:")
print(df5_4_grouped_product)
```
