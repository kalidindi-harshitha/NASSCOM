```python
a = np.array([1, 2, 3])
b = np.array([4, 0, 1])

# Dot product: sum of element-wise products -> measures alignment
print('a . b  (dot)   :', np.dot(a, b))      # 1*4 + 2*0 + 3*1 = 7

# Cross product: a new vector perpendicular to both (3-D only)
print('a x b  (cross) :', np.cross(a, b))

```

```python
print('L2 norm  ||a||_2 :', la.norm(a))         # sqrt(sum of squares)
print('L1 norm  ||a||_1 :', la.norm(a, 1))      # sum of absolute values
print('Linf norm        :', la.norm(a, np.inf)) # max absolute value

# Cosine similarity: the angle between vectors, ignoring magnitude
def cosine(u, v):
    return np.dot(u, v) / (la.norm(u) * la.norm(v))

print('cosine(a, b)     :', round(cosine(a, b), 3))
```

```python
import numpy as np

pairs = [
    (np.array([1, 0, 1]), np.array([1, 0, 1])),    # pair 1
    (np.array([1, 2, 3]), np.array([3, 2, 1])),    # pair 2
    (np.array([2, 0, 0]), np.array([0, 5, 0])),    # pair 3
]

similarities = []

for i, (u, v) in enumerate(pairs, start=1):

    # 1. L1 and L2 norm of u
    l1_norm = np.linalg.norm(u, ord=1)
    l2_norm = np.linalg.norm(u, ord=2)

    # 2. Cosine similarity of u and v
    cosine_similarity = np.dot(u, v) / (
        np.linalg.norm(u) * np.linalg.norm(v)
    )

    similarities.append(cosine_similarity)

    print(f"\nPair {i}:")
    print("u =", u)
    print("v =", v)
    print("L1 norm of u:", l1_norm)
    print("L2 norm of u:", l2_norm)
    print("Cosine similarity:", cosine_similarity)


# 3. Most similar pair
most_similar_pair = np.argmax(similarities) + 1

print("\nMost similar pair =", most_similar_pair)
```

```python
A = np.array([[2., 1.],
              [1., 3.]])
B = np.array([[1., 0.],
              [4., 2.]])

print('A @ B (matrix multiply):\n', A @ B)
print('A.T   (transpose):\n', A.T)
print('inverse(A):\n', la.inv(A))
print('trace(A) = sum of diagonal:', np.trace(A))
```

```python
I = np.eye(2)                       # identity: 1s on the diagonal
print('Identity I:\n', I)
print('A @ inv(A) == I ?', np.allclose(A @ la.inv(A), I))

# Symmetric: A equals its own transpose
print('A symmetric?', np.allclose(A, A.T))

# Orthogonal: Q.T @ Q == I (a pure rotation/reflection)
theta = np.radians(30)
Q = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])
print('Q orthogonal?', np.allclose(Q.T @ Q, I))
```

```python
import numpy as np

M = np.array([[4., 2.],
              [2., 3.]])

P = np.array([[0., -1.],
              [1.,  0.]])   # a 90-degree rotation


# 1. Inverse and trace of M
M_inverse = np.linalg.inv(M)
M_trace = np.trace(M)

print("1. Inverse of M:")
print(M_inverse)

print("Trace of M:", M_trace)


# 2. M @ inv(M) == I ?
identity_result = M @ M_inverse

print("\n2. M @ inv(M):")
print(identity_result)

print("Is M @ inv(M) equal to I?",
      np.allclose(identity_result, np.eye(2)))


# 3. Is M symmetric? Is P orthogonal?

is_M_symmetric = np.allclose(M, M.T)

is_P_orthogonal = np.allclose(
    P.T @ P,
    np.eye(2)
)

print("\n3. Is M symmetric?", is_M_symmetric)
print("Is P orthogonal?", is_P_orthogonal)
```

```python
# A unit square defined by its 4 corners (each column is a point)
square = np.array([[0, 1, 1, 0],
                   [0, 0, 1, 1]], dtype=float)

# Scaling matrix: stretch x by 1.5, y by 0.5
S = np.array([[1.5, 0.0],
              [0.0, 0.5]])

# Rotation matrix: rotate by 30 degrees
t = np.radians(30)
R = np.array([[np.cos(t), -np.sin(t)],
              [np.sin(t),  np.cos(t)]])

scaled  = S @ square      # apply scaling
rotated = R @ square      # apply rotation
print('Rotated corners:\n', rotated)
```

```python
def close_loop(pts):
    # repeat the first point at the end so the polygon closes
    return np.hstack([pts, pts[:, :1]])

fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(*close_loop(square),  marker='o', label='original')
ax.plot(*close_loop(scaled),  marker='o', label='scaled')
ax.plot(*close_loop(rotated), marker='o', label='rotated 30 deg')
ax.axhline(0, color='gray', lw=0.5); ax.axvline(0, color='gray', lw=0.5)
ax.set_aspect('equal'); ax.legend(); ax.set_title('Linear transformations')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

tri = np.array([[0, 1, 0.5],
                [0, 0, 1.0]])   # 3 corners of a triangle


# 1. Scaling matrix (x*2, y*0.5)
S = np.array([
    [2, 0],
    [0, 0.5]
])


# 2. Rotation matrix (45 degrees)
theta = np.radians(45)

R = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta),  np.cos(theta)]
])


print("Scaling matrix:")
print(S)

print("\nRotation matrix:")
print(R)


# 3. Apply both transformations
# First scale, then rotate
transformed_tri = R @ S @ tri

print("\nOriginal triangle:")
print(tri)

print("\nTransformed triangle:")
print(transformed_tri)


# Plot original vs transformed
plt.figure(figsize=(8, 6))

# Original triangle
plt.plot(
    np.append(tri[0], tri[0, 0]),
    np.append(tri[1], tri[1, 0]),
    marker='o',
    label='Original'
)

# Transformed triangle
plt.plot(
    np.append(transformed_tri[0], transformed_tri[0, 0]),
    np.append(transformed_tri[1], transformed_tri[1, 0]),
    marker='o',
    label='Transformed'
)

plt.axhline(0)
plt.axvline(0)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Original vs Transformed Triangle")

plt.legend()
plt.grid(True)

plt.axis('equal')
plt.show()
```

```python

A = np.array([[2., 0.],
              [0., 3.]])

vals, vecs = la.eig(A)     # vals = eigenvalues, vecs columns = eigenvectors
print('Eigenvalues  (lambda):', vals)
print('Eigenvectors (columns):\n', vecs)

```

```python
for i in range(len(vals)):
    v = vecs[:, i]                 # i-th eigenvector
    lhs = A @ v                    # A v
    rhs = vals[i] * v              # lambda v
    print(f'lambda={vals[i]:.1f}  A v == lambda v ?', np.allclose(lhs, rhs))
```

```python
import numpy as np

C = np.array([[4., 1.],
              [2., 3.]])


# 1. Eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(C)


# 2. Print eigenvalues
print("Eigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)


# 3. Verify C v = lambda v for the first eigenvector

lam = eigenvalues[0]
v = eigenvectors[:, 0]

left_side = C @ v
right_side = lam * v

print("\nFirst eigenvalue:", lam)
print("First eigenvector:", v)

print("\nC @ v:")
print(left_side)

print("\nlambda * v:")
print(right_side)

print("\nIs C v = lambda v?",
      np.allclose(left_side, right_side))
```

```python

A = np.array([[ 2.,  1., -1.],
              [-3., -1.,  2.],
              [-2.,  1.,  2.]])
b = np.array([8., -11., -3.])

x = la.solve(A, b)
print('Solution x:', x)                    # [2, 3, -1]
print('Rank of A :', la.matrix_rank(A))     # 3 -> full rank
print('Full rank -> a unique solution exists')
```

```python
import numpy as np

A2 = np.array([[3., 2.],
               [1., 4.]])

b2 = np.array([7., 9.])

cat = np.array([0.9, 0.8, 0.1])
dog = np.array([0.85, 0.7, 0.2])
car = np.array([0.1, 0.2, 0.95])


# 1. Solve A2 x = b2
x = np.linalg.solve(A2, b2)

print("1. Solution of A2 x = b2:")
print(x)


# 2. Rank of A2
rank_A2 = np.linalg.matrix_rank(A2)

print("\n2. Rank of A2:")
print(rank_A2)


# 3. Cosine similarity
def cosine_similarity(a, b):
    return np.dot(a, b) / (
        np.linalg.norm(a) * np.linalg.norm(b)
    )


cat_dog = cosine_similarity(cat, dog)
cat_car = cosine_similarity(cat, car)

print("\n3. Cosine similarity:")
print("Cat-Dog:", cat_dog)
print("Cat-Car:", cat_car)


# Which is more similar?
if cat_dog > cat_car:
    print("Cat is more similar to Dog.")
else:
    print("Cat is more similar to Car.")
```
