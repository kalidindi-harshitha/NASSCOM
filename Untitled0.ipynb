```python
 import os
import datetime
import random

print("--- Intermediate Python Core Exercise ---")
print("This exercise covers Python fundamentals, data structures, functions,")
print("core libraries (os, datetime, random), file handling, and exception handling.")
print("Estimated completion time: 3-4 hours.")
print("------------------------------------------")


# --- Global Data for the Exercise (Simulated Data) ---

DEVICE_TYPES = [
    'Temperature_Sensor',
    'Humidity_Sensor',
    'Pressure_Sensor',
    'Light_Sensor'
]

LOCATIONS = [
    'Lab1',
    'Lab2',
    'OfficeA',
    'OfficeB',
    'Warehouse'
]

READING_RANGES = {
    'Temperature_Sensor': (18.0, 30.0),
    'Humidity_Sensor': (30.0, 80.0),
    'Pressure_Sensor': (980.0, 1030.0),
    'Light_Sensor': (100, 1000)
}

```

```python
print("\n--- Task 1: Data Generation and Basic Structures ---")


# ------------------------------------------------------------
# 1.1 Generate 100 simulated sensor readings
# ------------------------------------------------------------

raw_sensor_data = []

start_time = datetime.datetime.now()

for i in range(100):

    timestamp = start_time + datetime.timedelta(minutes=i)

    timestamp_str = timestamp.strftime('%Y-%m-%d %H:%M:%S')

    device_id = f"DEV{i + 1:03d}"

    device_type = random.choice(DEVICE_TYPES)

    location = random.choice(LOCATIONS)

    min_value, max_value = READING_RANGES[device_type]

    value = round(random.uniform(min_value, max_value), 2)

    reading = (
        timestamp_str,
        device_id,
        device_type,
        location,
        value
    )

    raw_sensor_data.append(reading)


print("\n1.1 Total readings generated:", len(raw_sensor_data))
print("First reading:", raw_sensor_data[0])
print("Last reading:", raw_sensor_data[-1])


# ------------------------------------------------------------
# 1.2 Create temperature_readings list of dictionaries
# ------------------------------------------------------------

temperature_readings = []

for reading in raw_sensor_data:

    timestamp = reading[0]
    device_id = reading[1]
    device_type = reading[2]
    location = reading[3]
    value = reading[4]

    if device_type == 'Temperature_Sensor':

        temperature_dict = {
            'timestamp': timestamp,
            'device_id': device_id,
            'location': location,
            'temperature_celsius': value
        }

        temperature_readings.append(temperature_dict)


print("\n1.2 Number of temperature readings:",
      len(temperature_readings))

if temperature_readings:
    print("First temperature reading:",
          temperature_readings[0])


# ------------------------------------------------------------
# 1.3 Create unique_locations set
# ------------------------------------------------------------

unique_locations = set()

for reading in raw_sensor_data:
    unique_locations.add(reading[3])


print("\n1.3 Unique locations:")
print(unique_locations)


# ------------------------------------------------------------
# 1.4 Create device_stats dictionary
# ------------------------------------------------------------

device_stats = {}

for device_type in DEVICE_TYPES:

    min_value = float('inf')
    max_value = float('-inf')

    for reading in raw_sensor_data:

        if reading[2] == device_type:

            value = reading[4]

            if value < min_value:
                min_value = value

            if value > max_value:
                max_value = value

    if min_value != float('inf'):
        device_stats[device_type] = (
            min_value,
            max_value
        )


print("\n1.4 Device statistics:")

for device_type, stats in device_stats.items():

    print(
        device_type,
        "-> Min:",
        stats[0],
        "Max:",
        stats[1]
    )


# ============================================================
# TASK 1.5: Advanced Indexing and Slicing
# ============================================================

print("\n--- Task 1.5: Advanced Indexing and Slicing (Lists and Strings) ---")


sample_string = "PythonProgrammingIsFun"


# ------------------------------------------------------------
# 1.5.1 List Slicing
# ------------------------------------------------------------

# 10th to 20th reading inclusive
# Python indexes start from 0,
# so index 9 represents the 10th element
# and index 19 represents the 20th element.

sliced_readings_1 = raw_sensor_data[9:20]

# Every 5th reading starting from beginning
sliced_readings_2 = raw_sensor_data[::5]


print("\n1.5.1")
print("Readings 10th to 20th:")
print(sliced_readings_1)

print("\nEvery 5th reading:")
print(sliced_readings_2)


# ------------------------------------------------------------
# 1.5.2 String Slicing
# ------------------------------------------------------------

# "PythonProgrammingIsFun"
# Python = indexes 0 to 5
# Programming = indexes 6 to 16

substring_1 = sample_string[6:17]

# Reverse "Python"
# Python = sample_string[0:6]
# Reverse using slicing

substring_2 = sample_string[0:6][::-1]


print("\n1.5.2")
print("Substring 1:", substring_1)
print("Substring 2:", substring_2)


# ------------------------------------------------------------
# 1.5.3 Mixed Slicing
# ------------------------------------------------------------

# 5th reading -> index 4
fifth_reading_device_type = raw_sensor_data[4][2]

# Last 6 characters
device_type_suffix = fifth_reading_device_type[-6:]


print("\n1.5.3")
print("5th reading device type:",
      fifth_reading_device_type)

print("Last 6 characters:",
      device_type_suffix)
```

```python
print("\n--- Task 2: Function Definitions and Usage ---")


# ------------------------------------------------------------
# 2.1 get_readings_by_location()
# ------------------------------------------------------------

def get_readings_by_location(readings_list, target_location):

    result = []

    for reading in readings_list:

        if reading[3] == target_location:
            result.append(reading)

    return result


# Test function
lab1_readings = get_readings_by_location(
    raw_sensor_data,
    'Lab1'
)

print("\n2.1 Number of readings in Lab1:",
      len(lab1_readings))


# ------------------------------------------------------------
# 2.2 calculate_average_temperature()
# ------------------------------------------------------------

def calculate_average_temperature(temp_data_list):

    if len(temp_data_list) == 0:
        return 0.0

    total = 0

    for reading in temp_data_list:
        total += reading['temperature_celsius']

    average = total / len(temp_data_list)

    return average


# Test function
average_temperature = calculate_average_temperature(
    temperature_readings
)

print("\n2.2 Average temperature:",
      round(average_temperature, 2),
      "°C")


# ------------------------------------------------------------
# 2.3 generate_report_line()
# ------------------------------------------------------------

def generate_report_line(reading_dict):

    return (
        f"[{reading_dict['timestamp']}] "
        f"Device {reading_dict['device_id']} "
        f"at {reading_dict['location']}: "
        f"{reading_dict['temperature_celsius']}°C"
    )


# Test function
if temperature_readings:

    sample_report = generate_report_line(
        temperature_readings[0]
    )

    print("\n2.3 Sample report line:")
    print(sample_report)
```

```python
print("\n--- Task 3: Core Library Usage ---")


# ------------------------------------------------------------
# 3.1 Create sensor_reports directory
# ------------------------------------------------------------

reports_dir = 'sensor_reports'

if not os.path.exists(reports_dir):

    os.makedirs(reports_dir)

    print(f"\nDirectory '{reports_dir}' created.")

else:

    print(f"\nDirectory '{reports_dir}' already exists.")


# ------------------------------------------------------------
# 3.2 Calculate duration
# ------------------------------------------------------------

if len(raw_sensor_data) > 1:

    first_timestamp_str = raw_sensor_data[0][0]

    last_timestamp_str = raw_sensor_data[-1][0]

    date_format = '%Y-%m-%d %H:%M:%S'

    first_dt = datetime.datetime.strptime(
        first_timestamp_str,
        date_format
    )

    last_dt = datetime.datetime.strptime(
        last_timestamp_str,
        date_format
    )

    duration = last_dt - first_dt

    days = duration.days

    total_seconds = int(duration.total_seconds())

    hours = (total_seconds // 3600) % 24

    minutes = (total_seconds % 3600) // 60

    print(
        f"\nDuration of data collection: "
        f"{days} days, "
        f"{hours} hours, "
        f"{minutes} minutes."
    )

else:

    print("\nNot enough data to calculate duration.")


# ------------------------------------------------------------
# 3.3 Generate 5 random integers
# ------------------------------------------------------------

random_values = []

for i in range(5):

    random_values.append(
        random.randint(100, 200)
    )


print("\n3.3 Random values:")
print(random_values)
```

```python

print("\n--- Task 4: File I/O ---")


# ------------------------------------------------------------
# 4.1 Write temperature readings to file
# ------------------------------------------------------------

report_filename = os.path.join(
    reports_dir,
    'temperature_report.txt'
)

try:

    with open(report_filename, 'w') as f:

        for reading_dict in temperature_readings:

            line = generate_report_line(
                reading_dict
            )

            f.write(line + '\n')

    print(
        f"\nTemperature report written to "
        f"'{report_filename}'."
    )

except IOError as e:

    print(
        f"\nError writing file "
        f"'{report_filename}': {e}"
    )


# ------------------------------------------------------------
# 4.2 Read report file
# ------------------------------------------------------------

read_report_lines = []

try:

    with open(report_filename, 'r') as f:

        read_report_lines = f.readlines()

    print("\n4.2 First 3 lines of the report:")

    for line in read_report_lines[:3]:

        print(line.strip())

except FileNotFoundError:

    print(
        f"\nFile '{report_filename}' was not found."
    )

except IOError as e:

    print(
        f"\nError reading file: {e}"
    )


# ============================================================
# EXERCISE COMPLETE
# ============================================================

print("\n--- Exercise Complete! ---")

print(
    "Great job completing the intermediate Python exercise."
)

print(
    "Review your code and outputs to ensure correctness "
    "and understanding of all concepts."
)

print(
    "Consider adding more complex scenarios or "
    "user inputs for further practice."
)
```
